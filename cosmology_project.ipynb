# Cosmological Simulation in ΛCDM Model

This project explores the expansion of the universe using the ΛCDM cosmological model.

The goal is to analyze how different cosmological parameters affect the evolution of the universe, including the Hubble parameter, scale factor, and cosmic distances.

```python
import numpy as np
import matplotlib.pyplot as plt

# Cosmological parameters
H0 = 70  # Hubble constant (km/s/Mpc)
Omega_m = 0.3
Omega_lambda = 0.7

# Redshift range
z = np.linspace(0, 5, 500)

# Hubble parameter function
def H(z):
    return H0 * np.sqrt(Omega_m * (1 + z)**3 + Omega_lambda)

# Scale factor
a = 1 / (1 + z)

# Plot H(z)
plt.figure(figsize=(4, 4))
plt.plot(z, H(z))
plt.xlabel("Redshift (z)")
plt.ylabel("H(z)")
plt.title("Hubble Parameter vs Redshift")
plt.grid()
plt.show()

# Plot scale factor
plt.figure(figsize=(4, 4))
plt.plot(z, a)
plt.xlabel("Redshift (z)")
plt.ylabel("Scale Factor a(z)")
plt.title("Scale Factor vs Redshift")
plt.grid()
plt.show()
```

```python
# Speed of light (km/s)
c = 3e5

# Comoving distance function
def comoving_distance(z_vals):
    dz = z_vals[1] - z_vals[0]
    integral = np.cumsum(c / H(z_vals)) * dz
    return integral

# Calculate distance
d_c = comoving_distance(z)

# Plot
plt.figure(figsize=(5, 4))
plt.plot(z, d_c)
plt.xlabel("Redshift (z)")
plt.ylabel("Comoving Distance (Mpc)")
plt.title("Comoving Distance vs Redshift")
plt.grid()
plt.show()
```

```python
# Define different cosmological models
models = {
    "Matter Dominated (Ωm=1)": (1.0, 0.0),
    "Dark Energy Dominated (ΩΛ=1)": (0.0, 1.0),
    "ΛCDM (Ωm=0.3, ΩΛ=0.7)": (0.3, 0.7)
}

# Updated Hubble function
def H_model(z, Om, Ol):
    return H0 * np.sqrt(Om * (1 + z)**3 + Ol)

# Plot H(z) for different models
plt.figure(figsize=(5, 4))
for name, (Om, Ol) in models.items():
    plt.plot(z, H_model(z, Om, Ol), label=name)

plt.xlabel("Redshift (z)")
plt.ylabel("H(z)")
plt.title("Comparison of Cosmological Models")
plt.legend()
plt.grid()
plt.show()

# Plot Comoving Distance for different models
plt.figure(figsize=(5, 4))
for name, (Om, Ol) in models.items():
    dz = z[1] - z[0]
    d = np.cumsum(3e5 / H_model(z, Om, Ol)) * dz
    plt.plot(z, d, label=name)

plt.xlabel("Redshift (z)")
plt.ylabel("Comoving Distance (Mpc)")
plt.title("Distance Comparison Between Models")
plt.legend()
plt.grid()
plt.show()
```

## Analysis of Results

The comparison between different cosmological models shows that the expansion history of the universe strongly depends on the relative contributions of matter and dark energy.

In the matter-dominated model, the expansion rate decreases more rapidly over time, while in the dark energy-dominated model, the expansion accelerates significantly.

The ΛCDM model provides a balanced behavior, consistent with current cosmological observations, where dark energy drives late-time acceleration while matter dominates early structure formation.

These results demonstrate how cosmological parameters influence the large-scale evolution of the universe.
